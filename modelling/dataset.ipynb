# Shakespeare character dataset

Create the normalized character-level dataset from `shakespeare.txt` and inspect a few examples.

```python
from pathlib import Path
import sys

repo_root = Path.cwd()
if not (repo_root / "modelling").exists():
    repo_root = repo_root.parent

sys.path.insert(0, str(repo_root))

from modelling.shakespeare import build_datasets
```

```python
block_size = 64
train_fraction = 0.9

train_dataset, test_dataset, vocab = build_datasets(
    block_size=block_size,
    train_fraction=train_fraction,
)

print(f"vocab size: {vocab.size}")
print(f"train samples: {len(train_dataset)}")
print(f"test samples: {len(test_dataset)}")
print(f"vocab {''.join(vocab.itos)}")
```

```python
examples = []
for idx in range(5):
    x, y = train_dataset[idx]
    examples.append(
        {
            "index": idx,
            "input_ids": x.tolist(),
            "label_ids": y.tolist(),
            "input_text": vocab.decode(x.tolist()),
            "label_text": vocab.decode(y.tolist()),
        }
    )

examples
```
